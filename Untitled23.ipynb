```python
# Import libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Upload dataset in Colab
from google.colab import files
uploaded = files.upload()

# Load dataset
df = pd.read_csv("/content/DailyDelhiClimateTest.csv")

# Convert date column to datetime
df['date'] = pd.to_datetime(df['date'])

# Set date as index
df.set_index('date', inplace=True)

# -----------------------------
# 1. Line Graph (Temperature Trend)
# -----------------------------
plt.figure(figsize=(10,5))
plt.plot(df.index, df['meantemp'], label='Mean Temperature')
plt.title("Temperature Trend Over Time")
plt.xlabel("Date")
plt.ylabel("Temperature")
plt.legend()
plt.show()
```

```python
# -----------------------------
# 2. Trend Line (Using Seaborn)
# -----------------------------
plt.figure(figsize=(10,5))
sns.regplot(x=df.index.map(pd.Timestamp.toordinal), y=df['meantemp'], scatter=False)
plt.title("Temperature Trend Line")
plt.xlabel("Time")
plt.ylabel("Temperature")
plt.show()
```

```python
# -----------------------------
# 3. Area Chart (Humidity Variation)
# -----------------------------
plt.figure(figsize=(10,5))
plt.fill_between(df.index, df['humidity'], color='skyblue', alpha=0.5)
plt.title("Humidity Variation Over Time")
plt.xlabel("Date")
plt.ylabel("Humidity")
plt.show()
```

```python
# -----------------------------
# 4. Line Graph (Wind Speed)
# -----------------------------
plt.figure(figsize=(10,5))
plt.plot(df.index, df['wind_speed'], color='green')
plt.title("Wind Speed Over Time")
plt.xlabel("Date")
plt.ylabel("Wind Speed")
plt.show()
```

```python
# -----------------------------
# 5. Multiple Climate Variables
# -----------------------------
plt.figure(figsize=(12,6))
plt.plot(df.index, df['meantemp'], label='Temperature')
plt.plot(df.index, df['humidity'], label='Humidity')
plt.plot(df.index, df['wind_speed'], label='Wind Speed')
plt.plot(df.index, df['meanpressure'], label='Pressure')

plt.title("Climate Variables Over Time")
plt.xlabel("Date")
plt.ylabel("Values")
plt.legend()
plt.show()
```
